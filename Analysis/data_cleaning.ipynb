```python
# Load libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os

RNG_SEED = 8
np.random.seed(RNG_SEED)
curr_dir = os.getcwd()
DATA_PATH = os.path.join(os.path.dirname(curr_dir), 'Data')
RESULTS_PATH = os.path.join(os.path.dirname(curr_dir), 'Results')
```

```python
print("DATA_PATH =", DATA_PATH)
```

```python
df =  pd.read_csv(os.path.join(DATA_PATH, "machine_learning_data_template_ca125 serum.csv"))

cols_to_avg = ['Rsol', 'Rp', 'CPE', 'CV_Area', 'Delta_Z']

# Build aggregation: mean for numeric cols, first for Class_Multimodal
agg_dict = {col: 'mean' for col in cols_to_avg}
agg_dict['Class_Multimodal'] = 'first'

# Group by 'Type' and aggregate
df_avg = df.groupby('Type').agg(agg_dict).reset_index()

# Reorder columns: Type, Class_Multimodal, then averaged metrics
df_avg = df_avg[['Type', 'Class_Multimodal'] + cols_to_avg]



# Sort by that new column, then drop it
df_avg = df_avg.sort_values('conc').drop(columns=['conc']).reset_index(drop=True)


# Extract the integer from the "Type" column and create a new column "Type_float (ng/ml)"
# This assumes the number is the first occurrence of digits in the string.
df_avg['Type_float (ng/ml)'] = df_avg['Type'].str.extract(r'(\d+)').astype(float)

cols = df_avg.columns.tolist()
cols.remove("Type_float (ng/ml)")
new_cols = [cols[0], "Type_float (ng/ml)"] + cols[1:]
df_avg = df_avg[new_cols]

# Now df_avg is ordered from lowest to highest concentration and 
print(df_avg)
print("Data types by column:")
print(df_avg.dtypes)

# Display the result
df_avg
```
