# Set Up

## Imports and 

```python
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

## Seeds and Paths

```python
RNG_SEED = 8
np.random.seed(seed=RNG_SEED)
curr_dir = os.getcwd()
DATA_PATH = os.path.join(os.path.dirname(curr_dir), 'Data')
MODELS_PATH = os.path.join(os.path.dirname(curr_dir), 'Models')
WEIGHTS_PATH = os.path.join(os.path.dirname(curr_dir), 'Weights')
ASSETS_PATH = os.path.join(os.path.dirname(curr_dir), 'Assets')
RESULTS_PATH = os.path.join(os.path.dirname(curr_dir), 'Results')
```

# Splitting Dataset into train and validation sets

```python
ca_125_dataset = pd.read_csv(DATA_PATH + "/averaged_machine_learning_data_template_ca125 serum.csv")
```

```python
# Define the proportional size of the dataset split
# 80% train, 20% test split
df_train, df_test = train_test_split(
    ca_125_dataset,
    test_size=0.2,
    random_state=RNG_SEED,   # change or remove for different shuffles
    shuffle=True
)
# Check the splits with the line below
# print("train:\n", df_train, "\n\ntest:\n", df_test)
```

```python
# Save the datasets into train and test subsets in the folder Data
df_train.to_csv(DATA_PATH + "/ca_125_train.csv", index=False)
df_test.to_csv(DATA_PATH + "/ca_125_test.csv", index=False)
```
