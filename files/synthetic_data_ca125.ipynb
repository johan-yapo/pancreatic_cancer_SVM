# Imports

```python
import pandas as pd
import numpy as np
import os
from sdv.single_table import CTGANSynthesizer
from sdv.metadata import SingleTableMetadata
from sdv.metadata import Metadata

RNG_SEED = 8
np.random.seed(RNG_SEED)
curr_dir = os.getcwd()
DATA_PATH = os.path.join(os.path.dirname(curr_dir), 'Data')
RESULTS_PATH = os.path.join(os.path.dirname(curr_dir), 'Results')
```

# Load ca125 dataset

```python
# Load real dataset
real_data = pd.read_csv(os.path.join(DATA_PATH, "Vash_data/machine_learning_data_template_crp_serum.csv"))

TARGET = "Class_Multimodal" 

#Clean the target variable by stripping whitespace and converting to string
real_data[TARGET] = (
    real_data[TARGET]
    .astype(str)
    .str.strip()
)

print(TARGET)
```

Generating synthetic data

```python

# Define metadata

from sdv.metadata import Metadata
metadata = Metadata.detect_from_dataframe(real_data)


# Initialize and train CTGAN synthesizer
synth = CTGANSynthesizer(metadata, epochs=300, batch_size=512, pac=1)
synth.fit(real_data)

n_normal = 6
n_low = 6
n_high = 6 

# Generate synthetic normal risk class data
syn_norm = synth.sample_from_conditions(
    conditions=pd.DataFrame([{TARGET: "Normal"}]*n_normal)
)

syn_low = synth.sample_from_conditions(
    conditions=pd.DataFrame([{TARGET: "Low Risk"}]*n_low)
)

syn_high = synth.sample_from_conditions(
    conditions=pd.DataFrame([{TARGET: "High Risk"}]*n_high)
)

df_syn = pd.concat([syn_norm, syn_low, syn_high], ignore_index=True)

df_train_aug = pd.concat([real_data, df_syn], ignore_index=True)
```

```python
# Load real dataset
real_data = pd.read_csv(os.path.join(DATA_PATH, "Vash_data\\machine_learning_data_template_crp_serum.csv"))

TARGET = "Class_Multimodal" 

#Clean the target variable by stripping whitespace and converting to string
real_data[TARGET] = (
    real_data[TARGET]
    .astype(str)
    .str.strip()
)

print(TARGET)

from sdv.sampling import Condition
import pandas as pd
real_data[TARGET] = real_data[TARGET].str.strip()

class_counts = {
    "Normal": 6,
    "Low Risk": 6,
    "High Risk": 6
}

synthetic_parts = []

for cls, n in class_counts.items():
    syn_part = synth.sample_from_conditions(
        conditions=[Condition(num_rows=n, column_values={TARGET: cls})]
    )
    synthetic_parts.append(syn_part)

# Combine all synthetic data
df_syn = pd.concat(synthetic_parts, ignore_index=True)
df_syn.to_csv("../outputs/df_syn.csv", index=False)

# Merge with original dataset
df_train_aug = pd.concat([real_data, df_syn], ignore_index=True)
df_train_aug.to_csv("../Data/Vash_data/SyntheticData/Synthetically_Enhanced_machine_learning_data_template_crp_serum.csv", index=False)
```
