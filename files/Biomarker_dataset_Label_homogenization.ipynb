```python
import os
import pandas as pd
import numpy as np

RNG_SEED = 8
np.random.seed(RNG_SEED)
curr_dir = os.getcwd()
DATA_PATH = os.path.join(os.path.dirname(curr_dir), 'Data')
RESULTS_PATH = os.path.join(os.path.dirname(curr_dir), 'Results')
```

# CA19-9 Biomarker

```python
import pandas as pd
import re

# Load real CA dataset
df = pd.read_csv(os.path.join(DATA_PATH, "CA19-9 serum machine learning.csv"))

def convert_type(val):
    val = val.lower().strip()

    if val == "control":
        return "0"

    # Extract number + unit
    match = re.search(r'(\d+)(pg|ng)', val)
    if match:
        num = float(match.group(1))
        unit = match.group(2)

        if unit == "pg":
            num = num / 1000  # convert pg → ng

        return f"{num} ng/ml"

    return val  # fallback

df["Type"] = df["Type"].apply(convert_type)
```

```python
# Save cleaned file
df.to_csv(os.path.join(DATA_PATH, "cleaned_CA19-9_data.csv"), index=False)

```

# CEA Serum Biomarker

```python
df2 = pd.read_csv(os.path.join(DATA_PATH, "CEA serum machine learning.csv"))
print(df2["Type"].unique()[:20])

def convert_type(val):
    val = str(val).lower().strip()

    # ✅ If already converted → return as-is
    if "ng/ml" in val or val == "0":
        return val

    if val == "control":
        return "0"

    # Extract number + unit
    match = re.search(r'(\d+\.?\d*)(pg|ng)', val)

    if match:
        num = float(match.group(1))
        unit = match.group(2)

        if unit == "pg":
            num = num / 1000

        if num == 0:
            return "0"
        elif num.is_integer():
            return f"{int(num)} ng/ml"
        else:
            return f"{num:.3f} ng/ml"

    return val
df2["Type"] = df2["Type"].apply(convert_type)

for val in df2["Type"].unique():
    print(val, "→", convert_type(val))
```

```python
df2.to_csv(os.path.join(DATA_PATH, "cleaned_CEA_data.csv"), index=False)
```
